```python
import h2o
h2o.init(max_mem_size_GB=4, ip="127.0.0.1", port="8080")
```

```python
train = h2o.import_file("./data/train.csv")
test = h2o.import_file("./data/test.csv")
```

```python
x = train.columns
y = "Survived"
train[y] = train[y].asfactor()
x.remove(y)
train.describe()
```

```python
from h2o.automl import H2OAutoML
aml = H2OAutoML(max_models=10, seed=42, max_runtime_secs=7200, sort_metric="accuracy")
aml.train(x=x, y=y, training_frame=train)
```

```python
lb = aml.leaderboard
print(lb.head(rows=lb.nrows))
```
